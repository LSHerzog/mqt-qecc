```python
from mqt.qecc.co3.utils.lattice_router import plot_lattice_paths

layout = res_lst[0]["final_layout_lst"][0]
circuit = res_lst[0]["circuits"][0]
factory_pos = []
g, data_qubit_locs, factory_ring = co.plots.gen_layout("hex", 24, factory_pos)

m,n = 10,10
terminal_pairs = co.translate_layout_circuit(circuit, layout)

router = co.ShortestFirstRouterTGatesDyn(m,n, terminal_pairs, factory_pos, 0)
router.G = g #update g
```

```python
layers = router.find_total_vdp_layers_dyn()
```

```python
layout_red = {k: v for k, v in layout.items() if k != 'factory_positions'}
for i in range(10):
    if len(layers[i])!=0:
        print(f"----------i={i}---------")
        plot_lattice_paths(g = router.G, vdp_dict = layers[i], factory_locs=factory_pos, layout = layout_red, size=(8,8))
```
