```python
import matplotlib.pyplot as plt
import numpy as np

import mqt.qecc.co3 as co

plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["mathtext.fontset"] = "stix"

import pickle
from pathlib import Path
```

```python
path = "./results/f_vs_time_q42_ratio08_small_row_FSC"

with Path(path).open("rb") as f:
    res_lst = pickle.load(f)

path = "./results/f_vs_time_q42_ratio08_small_row_FSC_metricrouting"

with Path(path).open("rb") as f:
    res_lst_routing = pickle.load(f)
```

```python
#split the lists into the contributions from fSC and CC graphs
res_lst_fsc = []
res_lst_cc = []
for i, res in enumerate(res_lst):
    if res["instances"][i]["graphtype"] == "FSC":
        res_lst_fsc.append(res)
    elif res["instances"][i]["graphtype"] == "CC":
        res_lst_cc.append(res)
    else:
        raise ValueError("incorrect graphtype")
    

res_lst_routing_fsc = []
res_lst_routing_cc = []
for i, res in enumerate(res_lst_routing):
    if res["instances"][i]["graphtype"] == "FSC":
        res_lst_routing_fsc.append(res)
    elif res["instances"][i]["graphtype"] == "CC":
        res_lst_routing_cc.append(res)
    else:
        raise ValueError("incorrect graphtype")
```

```python
print(len(res_lst_routing_cc))
print(len(res_lst_routing_fsc))
print(len(res_lst_cc))
print(len(res_lst_fsc))
```

```python
res_lst = res_lst_fsc
instances = res_lst[0]["instances"]
hc_params = res_lst[0]["hc_params"]
size = (2.5,2)
q=instances[0]["q"]
ratio=instances[0]["ratio"]
layout_name = instances[0]["layout_name"]
min_depth = instances[0]["min_depth"]
graphtype = "FSC"
co.plots.plot_f_vs_t(res_lst, q, ratio, layout_name, min_depth, graphtype, hc_params, size=size)
```
